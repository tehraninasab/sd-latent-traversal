```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# List of files to process
files = ['cosine_similarity_sd.csv', 'cosine_similarity_pe.csv']

# Set up the plot
plt.figure(figsize=(5, 5))

# Markers and colors for different datasets
markers = ['o', 'd', 's', '^']  # Add more markers if needed
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']  # Add more colors if needed

# Process each file
for idx, file in enumerate(files):
    # Read the CSV
    df = pd.read_csv('/usr/local/data/amarkr/dent/metrics/'+file)
    
    # Get step columns
    step_columns = [col for col in df.columns if col.startswith('Step')]
    
    # Calculate mean and std for each step
    means = df[step_columns].mean()
    stds = df[step_columns].std()
    
    # Extract step numbers for x-axis
    steps = [int(col.split()[1]) for col in step_columns]
    
    # Create label from filename (remove prefix and extension)
    label = file.replace('cosine_similarity_', '').replace('.csv', '').upper()
    
    # Plot mean line with error bars
    plt.errorbar(steps, means, yerr=stds, fmt=f'{markers[idx]}-', 
                capsize=5, capthick=1.5, elinewidth=1.5, markersize=8,
                color=colors[idx], label=f'{label}: Mean ± Std Dev')

# Customize the plot
plt.xlabel('T', fontsize=16)
plt.ylabel('Cosine Similarity', fontsize=16)
plt.legend(fontsize=10)

# Add minor gridlines
plt.minorticks_on()
plt.grid(True, which='minor', linestyle=':', alpha=0.4)

# Adjust layout
plt.tight_layout()

# Show the plot
plt.show()

# Print numerical values for each dataset
print("\nNumerical Values:")
for file in files:
    print(f"\n{file.replace('cosine_similarity_', '').replace('.csv', '').upper()}:")
    df = pd.read_csv('/usr/local/data/amarkr/dent/metrics/'+file)
    step_columns = [col for col in df.columns if col.startswith('Step')]
    means = df[step_columns].mean()
    stds = df[step_columns].std()
    steps = [int(col.split()[1]) for col in step_columns]
    
    for step, mean, std in zip(steps, means, stds):
        print(f"Step {step:2d}: Mean = {mean:.4f} ± {std:.4f}")
```
